## 2. Python fundamentals
Variables, types, collections, conditionals, loops, functions, and defensive handling of edge cases.

```python
candidate = "Ganesh"
hours = 4.5
completed = True
scores = [45, 48, 52, 56, 60]
profile = {"candidate": candidate, "hours": hours, "completed": completed}

print(type(candidate).__name__, type(hours).__name__, type(completed).__name__)
print("List length:", len(scores))
print("Dictionary candidate:", profile["candidate"])

def classify_score(score):
    if not isinstance(score, (int, float)):
        raise TypeError("score must be numeric")
    if score < 0 or score > 100:
        raise ValueError("score must be between 0 and 100")
    if score >= 85:
        return "High"
    elif score >= 60:
        return "Pass"
    return "Needs improvement"

for score in scores:
    print(score, classify_score(score))
```

## 3. Statistical worked example
For the small sample `[10, 12, 12, 14, 17]`, calculate center and spread. The variance and standard deviation below are **sample** statistics (denominator `n - 1`).

```python
sample = [10, 12, 12, 14, 17]
print("Mean:", np.mean(sample))
print("Median:", np.median(sample))
print("Mode:", pd.Series(sample).mode().tolist())
print("Range:", np.max(sample) - np.min(sample))
print("Sample variance:", np.var(sample, ddof=1))
print("Sample standard deviation:", np.std(sample, ddof=1))
```

## 4. Probability example
If 3 of 10 equally likely tickets win, the probability of selecting a winning ticket in one draw is `3/10 = 0.30` (30%). This assumes each ticket is equally likely and the draw is fair.

```python
winning_tickets = 3
total_tickets = 10
probability = winning_tickets / total_tickets
print(f"Probability: {probability:.2f} ({probability:.0%})")
```

## 5. Load and inspect the dataset
The raw CSV intentionally contains missing values and one duplicate row to demonstrate quality checks.

```python
raw = pd.read_csv("student_performance_raw.csv")
print("Shape:", raw.shape)
print("\nFirst five rows:")
display(raw.head())
print("\nData types:")
print(raw.dtypes)
print("\nMissing values by column:")
print(raw.isna().sum())
print("\nDuplicate rows:", raw.duplicated().sum())
```

## 6. Data cleaning decisions
- Remove exact duplicate rows.
- Fill missing numeric values with the median of that column for this small demonstration.
- In a real project, investigate why data are missing and assess whether imputation is justified before applying it.

```python
clean = raw.drop_duplicates().copy()
for col in ["Attendance_Pct", "Quiz_Score", "Final_Score"]:
    clean[col] = clean[col].fillna(clean[col].median())

# Validate expected ranges
assert clean["Attendance_Pct"].between(0, 100).all()
assert clean["Quiz_Score"].between(0, 100).all()
assert clean["Final_Score"].between(0, 100).all()

print("Cleaned shape:", clean.shape)
print("Remaining missing values:", int(clean.isna().sum().sum()))
print("Remaining duplicate rows:", int(clean.duplicated().sum()))
display(clean.describe().round(2))
```

## 7. EDA and original observations
The following cells compute the results from the supplied synthetic dataset. Treat these as demonstrations, not evidence about real students.

```python
print("Average final score:", round(clean["Final_Score"].mean(), 2))
print("Median final score:", round(clean["Final_Score"].median(), 2))
print("Students scoring at least 60:", int((clean["Final_Score"] >= 60).sum()), "of", len(clean))
print("Share scoring at least 60:", round((clean["Final_Score"] >= 60).mean() * 100, 1), "%")
print("Correlation: study hours vs final score:", round(clean["Study_Hours"].corr(clean["Final_Score"]), 3))
print("Correlation: attendance vs final score:", round(clean["Attendance_Pct"].corr(clean["Final_Score"]), 3))
```

```python
plt.figure(figsize=(7, 4.5))
plt.scatter(clean["Study_Hours"], clean["Final_Score"])
plt.xlabel("Study hours per week")
plt.ylabel("Final score")
plt.title("Study Hours vs Final Score (Synthetic Data)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(7, 4.5))
plt.hist(clean["Final_Score"], bins=6, edgecolor="black")
plt.xlabel("Final score")
plt.ylabel("Number of students")
plt.title("Distribution of Final Scores (Synthetic Data)")
plt.tight_layout()
plt.show()
```

## 8. Interpretation and limitations
1. The dataset is deliberately constructed so scores generally rise with study hours and attendance; correlations are therefore expected to be positive.
2. Correlation is an association, not proof that additional study hours or attendance alone caused higher scores.
3. The sample is small and synthetic, so findings must not be generalized to a real student population.
4. Median imputation is a teaching choice. In production analysis, document missingness patterns and compare alternative methods.
5. Before drawing business or policy conclusions, obtain representative real data, validate collection methods, and consider confounding variables.
